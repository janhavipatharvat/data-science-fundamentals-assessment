## 1. Data Science Fundamentals

### 1.1 What is Data Science?

### 2.2 Conditional statements

```python
def grade(score):
    """Return a letter grade for a numeric score."""
    if score >= 90:
        return "A"
    elif score >= 75:
        return "B"
    elif score >= 60:
        return "C"
    else:
        return "F"

print([grade(s) for s in (95, 80, 62, 40)])
```

### 2.4 Functions

```python
def describe(numbers, round_to=2):
    """Return (min, max, average) of a list of numbers."""
    if not numbers:
        raise ValueError("numbers must not be empty")
    return min(numbers), max(numbers), round(sum(numbers) / len(numbers), round_to)

print(describe([4, 8, 15, 16, 23, 42]))

# lambda and higher-order functions
double = lambda x: x * 2
print(list(map(double, [1, 2, 3])), list(filter(lambda x: x > 1, [1, 2, 3])))
```

## 3. Statistical Foundations

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statistics import mean, median, mode, variance, stdev

data = [12, 15, 15, 18, 20, 22, 22, 22, 25, 30, 95]   # 95 is an outlier
```

### 3.1 Descriptive statistics
| Measure | Meaning |
|---|---|
| Mean | Arithmetic average, sensitive to outliers |
| Median | Middle value, robust to outliers |
| Mode | Most frequent value |
| Variance | Average squared distance from the mean |
| Std. deviation | Square root of variance, same units as the data |
| Percentile | Value below which a given % of data falls |

```python
print("Mean               :", round(mean(data), 2))
print("Median             :", median(data))
print("Mode               :", mode(data))
print("Sample variance    :", round(variance(data), 2))
print("Sample std dev     :", round(stdev(data), 2))
print("Range              :", max(data) - min(data))

q1, q2, q3 = np.percentile(data, [25, 50, 75])
iqr = q3 - q1
print(f"Q1={q1}, Q2={q2}, Q3={q3}, IQR={iqr}")
print("90th percentile    :", np.percentile(data, 90))

# Outlier detection using 1.5 * IQR rule
lo, hi = q1 - 1.5 * iqr, q3 + 1.5 * iqr
print("Outliers           :", [x for x in data if x < lo or x > hi])
```

```python
# Effect of an outlier on mean vs median
clean = [x for x in data if x != 95]
print(f"With outlier    -> mean {mean(data):.2f}, median {median(data)}")
print(f"Without outlier -> mean {mean(clean):.2f}, median {median(clean)}")

fig, ax = plt.subplots(1, 2, figsize=(10, 3.5))
ax[0].hist(data, bins=8, edgecolor="black"); ax[0].set_title("Histogram")
ax[1].boxplot(data, vert=False);            ax[1].set_title("Box plot (outlier visible)")
plt.tight_layout(); plt.show()
```

**Interpretation:** a single extreme value (95) pulls the mean far above the median, so for skewed data the median is the better measure of the typical value.

### 3.2 Probability basics
- P(A) = favourable outcomes / total outcomes, with 0 ≤ P ≤ 1
- Complement: P(not A) = 1 − P(A)
- Independent events: P(A and B) = P(A) × P(B)
- Conditional: P(A | B) = P(A and B) / P(B)

```python
import random
from itertools import product

# Exact: two dice
outcomes = list(product(range(1, 7), repeat=2))
p_sum7 = sum(1 for a, b in outcomes if a + b == 7) / len(outcomes)
print("P(sum = 7) exact       :", round(p_sum7, 4))

# Conditional: P(sum = 7 | first die is even)
cond = [o for o in outcomes if o[0] % 2 == 0]
print("P(sum=7 | first even)  :", round(sum(1 for a, b in cond if a + b == 7) / len(cond), 4))

# Simulation (law of large numbers)
random.seed(42)
for n in (100, 10_000, 200_000):
    hits = sum(random.randint(1, 6) + random.randint(1, 6) == 7 for _ in range(n))
    print(f"Simulated n={n:>7}: {hits / n:.4f}")
```

### 3.3 Correlation

```python
rng = np.random.default_rng(42)
hours = rng.uniform(1, 10, 60)
score = 40 + 5 * hours + rng.normal(0, 6, 60)      # positive relationship
noise = rng.normal(50, 10, 60)                      # unrelated

df_c = pd.DataFrame({"study_hours": hours, "exam_score": score, "random_noise": noise})
print(df_c.corr().round(2))

fig, ax = plt.subplots(1, 2, figsize=(10, 3.5))
ax[0].scatter(df_c.study_hours, df_c.exam_score); ax[0].set_title("Strong positive correlation")
ax[1].scatter(df_c.study_hours, df_c.random_noise); ax[1].set_title("No correlation")
for a in ax: a.set_xlabel("study_hours")
plt.tight_layout(); plt.show()
```

**Interpretation:** Pearson *r* ranges from −1 to +1. Values near ±1 mean a strong linear relationship; near 0 means little linear relationship.
**Correlation does not imply causation**: a hidden third variable (a confounder) or coincidence can create correlation.

**Common structures:** list/tuple/dict/set (Python), NumPy array (homogeneous numeric), pandas Series (1-D labelled) and DataFrame (2-D table).

```python
df = pd.DataFrame({
    "customer_id": [101, 102, 103, 104, 105, 106],
    "age": [25, 34, np.nan, 45, 29, 52],
    "city": ["Mumbai", "Delhi", "Mumbai", "Pune", "Delhi", "Pune"],
    "satisfaction": ["low", "high", "medium", "high", "low", "medium"],
    "signup_date": ["2024-01-15", "2024-03-02", "2024-03-20", "2024-07-11", "2024-09-05", "2024-12-25"],
    "review": ["Great service!", "Too slow...", "Okay, not bad", "LOVED it", "Poor support", "Fine."],
    "monthly_spend": [1200.5, 850.0, 990.0, 2300.75, 640.0, 1800.0],
})
print(df.dtypes)
```

```python
# Convert each column to its proper type
df["signup_date"] = pd.to_datetime(df["signup_date"])
df["city"] = df["city"].astype("category")
df["satisfaction"] = pd.Categorical(df["satisfaction"],
                                    categories=["low", "medium", "high"], ordered=True)

# Datetime feature extraction
df["signup_month"] = df["signup_date"].dt.month_name()
df["signup_weekday"] = df["signup_date"].dt.day_name()

# Ordinal encoding (keeps order) vs one-hot encoding (nominal)
df["satisfaction_code"] = df["satisfaction"].cat.codes
city_dummies = pd.get_dummies(df["city"], prefix="city")

# Text: simple cleaning
df["review_clean"] = (df["review"].str.lower()
                      .str.replace(r"[^a-z\s]", "", regex=True).str.strip())
df["review_words"] = df["review_clean"].str.split().str.len()

print(df[["signup_date", "signup_month", "signup_weekday", "satisfaction", "satisfaction_code",
          "review_clean", "review_words"]])
print()
print(city_dummies)
print("\nOrdinal comparison works:", (df["satisfaction"] >= "medium").tolist())
```

```python
# Handling missing values
print("Missing before:", df["age"].isna().sum())
df["age"] = df["age"].fillna(df["age"].median())   # median is robust to outliers
print("Missing after :", df["age"].isna().sum())
print(df[["customer_id", "age"]].T)
```

**Key rules**
- Never compute a mean on a nominal category (e.g. a "city code").
- Ordinal data keeps its order; nominal data should not be given fake order.
- Always parse dates into real datetime objects before analysis.
- Decide on a missing-value strategy deliberately and document it.

```python
# Bad vs good style
# BAD
def f(d):
    return d[d.a > 5].b.mean()

# GOOD
def average_spend_of_high_value_customers(customers: pd.DataFrame, min_orders: int = 5) -> float:
    """Return the mean monthly spend of customers with more than `min_orders` orders."""
    high_value = customers[customers["orders"] > min_orders]
    return high_value["monthly_spend"].mean()

demo = pd.DataFrame({"orders": [2, 6, 9, 4], "monthly_spend": [500, 1500, 2100, 700]})
print(average_spend_of_high_value_customers(demo))
```

### Responsible interpretation checklist
1. **Sample size and representativeness:** are conclusions drawn from enough, unbiased data?
2. **Correlation vs causation:** don't claim a cause without evidence.
3. **Outliers and missing data:** state how they were treated.
4. **Visual honesty:** label axes, start bars at zero, avoid misleading scales.
5. **Privacy and fairness:** protect personal data and check for biased outcomes.
6. **Uncertainty:** report ranges or limitations, not just point estimates.

### Assignment 2 – Statistics from scratch

```python
scores = [56, 67, 67, 72, 78, 81, 85, 85, 85, 90, 98]

def mean_(xs): return sum(xs) / len(xs)
def median_(xs):
    s, n = sorted(xs), len(xs)
    return s[n // 2] if n % 2 else (s[n // 2 - 1] + s[n // 2]) / 2
def mode_(xs):
    counts = {x: xs.count(x) for x in set(xs)}
    top = max(counts.values())
    return sorted(x for x, c in counts.items() if c == top)
def var_(xs, sample=True):
    m = mean_(xs)
    return sum((x - m) ** 2 for x in xs) / (len(xs) - (1 if sample else 0))

print("Mean    :", round(mean_(scores), 2))
print("Median  :", median_(scores))
print("Mode    :", mode_(scores))
print("Variance:", round(var_(scores), 2))
print("Std dev :", round(var_(scores) ** 0.5, 2))

# verify against numpy
assert abs(mean_(scores) - np.mean(scores)) < 1e-9
assert abs(var_(scores) - np.var(scores, ddof=1)) < 1e-9
print("Manual results match NumPy ✔")
```

### Assignment 3 – Data types and manipulation with pandas

```python
rng = np.random.default_rng(SEED)
n = 200
sales = pd.DataFrame({
    "order_id": range(1, n + 1),
    "order_date": pd.date_range("2025-01-01", periods=n, freq="D"),
    "region": rng.choice(["North", "South", "East", "West"], n),
    "category": rng.choice(["Electronics", "Clothing", "Grocery"], n, p=[0.3, 0.3, 0.4]),
    "quantity": rng.integers(1, 10, n),
    "unit_price": rng.normal(500, 150, n).round(2),
})
# inject data-quality problems
sales.loc[rng.choice(n, 8, replace=False), "unit_price"] = np.nan
sales = pd.concat([sales, sales.iloc[:3]], ignore_index=True)   # 3 duplicate rows

print(sales.shape); print(sales.info())
```

```python
raw = sales.copy()                                   # keep raw data untouched

def clean_data(df: pd.DataFrame) -> pd.DataFrame:
    """Remove duplicates, impute missing prices with category median, add revenue."""
    df = df.drop_duplicates().copy()
    df["unit_price"] = df.groupby("category")["unit_price"].transform(lambda s: s.fillna(s.median()))
    df["revenue"] = df["quantity"] * df["unit_price"]
    df["month"] = df["order_date"].dt.to_period("M").astype(str)
    return df

clean = clean_data(raw)
print("Rows before/after:", len(raw), "/", len(clean))
print("Missing values left:", int(clean.isna().sum().sum()))
clean.head()
```

```python
# Analysis: filtering, grouping, aggregation, sorting
print("Revenue by region:")
print(clean.groupby("region")["revenue"].agg(["count", "sum", "mean"]).round(2).sort_values("sum", ascending=False))

print("\nRevenue by category:")
print(clean.groupby("category")["revenue"].sum().round(2).sort_values(ascending=False))

print("\nTop 5 orders:")
print(clean.nlargest(5, "revenue")[["order_id", "region", "category", "revenue"]])

pivot = clean.pivot_table(index="region", columns="category", values="revenue", aggfunc="sum").round(0)
print("\nPivot table:"); print(pivot)
```

### Assignment 4 – Basic analysis and visualisation

```python
monthly = clean.groupby("month")["revenue"].sum()

fig, ax = plt.subplots(1, 3, figsize=(15, 4))
monthly.plot(kind="line", marker="o", ax=ax[0], title="Monthly revenue")
ax[0].set_ylabel("Revenue"); ax[0].tick_params(axis="x", rotation=45)

clean.groupby("category")["revenue"].sum().plot(kind="bar", ax=ax[1], title="Revenue by category")
ax[1].set_ylabel("Revenue")

ax[2].hist(clean["revenue"], bins=20, edgecolor="black"); ax[2].set_title("Distribution of order revenue")
ax[2].set_xlabel("Revenue")
plt.tight_layout(); plt.show()

print("Skewness of revenue:", round(clean["revenue"].skew(), 2))
print("Correlation (quantity vs revenue):", round(clean["quantity"].corr(clean["revenue"]), 2))
```

### Key findings (analysis summary)
- After cleaning, duplicates were removed and missing prices were imputed with the category median, leaving no missing values.
- Revenue is summarised by region, category and month above; the revenue distribution is right-skewed, so the median describes a typical order better than the mean.
- Quantity and revenue are positively correlated, as expected because revenue = quantity × price.
- **Limitations:** the data is synthetic, so results illustrate technique rather than real business conclusions; imputation adds uncertainty.
